```python
# Importando as bibliotecas essenciais para análise de dados
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuração visual dos gráficos
sns.set_theme(style='whitegrid', palette='muted')
%matplotlib inline
```

## 1. Carregando os Dados
Vamos carregar a base de treinamento do Titanic (`train.csv`).

```python
df_treino = pd.read_csv('../data/train.csv')
print(f'Formato da base de treino: {df_treino.shape[0]} linhas e {df_treino.shape[1]} colunas')
df_treino.head()
```

## 2. Identificando Dados Faltantes (Nulos)
Algumas informações importantes podem estar faltando em alguns passageiros.

```python
nulos = df_treino.isnull().sum()
print('Quantidade de valores nulos por coluna:')
print(nulos[nulos > 0])
```

## 3. Quem teve mais chance de sobreviver? (Análise Visual)
Vamos testar algumas hipóteses clássicas sobre o naufrágio:

```python
# 1. Taxa de Sobrevivência por Sexo
fig, ax = plt.subplots(figsize=(6, 4))
sns.barplot(data=df_treino, x='Sex', y='Survived', hue='Sex', palette='Set2', legend=False, ax=ax)
ax.set_title('Taxa de Sobrevivência por Sexo', fontsize=12, fontweight='bold')
ax.set_ylabel('Proporção de Sobrevivência (0 a 1)')
ax.set_xticklabels(['Masculino', 'Feminino'])
plt.show()
```

```python
# 2. Taxa de Sobrevivência por Classe do Bilhete
fig, ax = plt.subplots(figsize=(6, 4))
sns.barplot(data=df_treino, x='Pclass', y='Survived', hue='Pclass', palette='crest', legend=False, ax=ax)
ax.set_title('Taxa de Sobrevivência por Classe', fontsize=12, fontweight='bold')
ax.set_ylabel('Proporção de Sobrevivência (0 a 1)')
ax.set_xlabel('Classe (1ª, 2ª, 3ª)')
plt.show()
```

## 4. Principais Aprendizados e Conclusões
1. **Mulheres** tiveram uma prioridade enorme nos botes: cerca de **74% das mulheres** sobreviveram, comparado a apenas ~**18% dos homens**.
2. **Passageiros da 1ª Classe** tiveram taxas de sobrevivência superiores a 60%, enquanto os da 3ª classe ficaram perto de 24%.
